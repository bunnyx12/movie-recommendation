```python
import pandas as pd
movies = pd.read_csv("movies_with_emotions.csv")
big_data = pd.read_csv("D:/Downloads/TMDB_movie_dataset_v11.csv")
```

```python
merged = movies.merge(big_data[["title", "backdrop_path"]], on=["title"], how="left")
merged
```

```python
merged = merged.drop_duplicates(subset="id", keep="first")
merged
```

```python
movies = merged.rename(columns={"backdrop_path": "poster"})
movies.to_csv("movies_with_posters.csv", index=False)
```
